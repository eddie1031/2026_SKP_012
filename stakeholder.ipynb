# 이해관계자(Stakeholder)

서비스 기획에서는 서비스와 관련된 사람들을 사용자(User), 고객(Customer), 이해관계자(Stakeholder)로 구분한다. 사용자는 서비스를 직접 쓰는사람, 고객은 비용르 내거나 도입을 결정하는사람, 이해관계자는 서비스의 결과에 영향을 주거나 영향을 받는 모든 사람과 조직을 의미. 사용자랑 고객은 이해관계자 안에 포함된다.

- 사용자: 누가 화면을 보고 버튼을 누르는가
- 고객: 누가 돈을 내고 도입을 결정하는가
- 이해관계자: 누가 영향을 주거나 받는가

## 관계지도(Relationship Map)

관계지도는 이해관계자를 점으로 그 사이에 오가는 정보/허가/요청을 화살표로 그린 그림이다. 선박 한 척이 부산항에 들어오기 까지 선사는 대리점에 입항 일정을 알리고, 대리점은 해수부에 입항신고하고, 관제는 입항을 허가하고 도산/예선을 연결하고 터미널은 선석을 배정한다.

관계지도에서는 화살표가 많이 모이거나 나가는 특정 점이 중요한데, 이 점을 허브(Hub)라고 일컫는다. 허브는 정보가 거쳐가는 길목이기때문에 그곳이 막히면 여러 사람들이 동시에 기다리게 된다.

그래서 특정 변수 때문에 일이 한꺼번에 몰리는 날에는 허브의 부담이 가장먼저 커지게된다. 점마다 들어오는 화살표(입력 차수) 수와 나가는 화살표(출력 차수) 수로 허브를 찾아서 파악할필요가 있다.

## 이해관계자 정리

이해관계자는 서비스의 결과에 영향을 주거나, 서비스 때문에 영향을 받는 모든 사람과 조직을 의미한다.

부산항만에는 선사, 터미널운영사, 화주(하주), 포워더, 부산항만공사, 관제(VTS), 해수부, 예/도선 업체, 육상운송사 등이 이해관계자에 포함될 수 있다.

이해관계자는 좋은 항만을 판단하는 기준과 숫자다 다르다. 이들이 차이가 무엇인지 선행으로 정리되어야 누가 누구에게 도움이되고 부담되는지를 파악할 수 있다.

### 이해관계자 KPI 충돌

이해관계자마다 관심있는 KPI를 나열하면 크게 두가지로 구분해볼 수 있다.

- 여러 이해관계자가 같이보는 KPI
- 여러 이해관계자가 서로 충돌하는 KPI

같이보는 KPI를 개선하는 서비스는 설득하기가 쉽다. 서로 충돌하는 KPI는 설득하기가 얼렵다.

터미널 운영사, 항만공사에게 선석 점유율이 높다 -> 시설을 효율적으로 잘 쓰고있다.
선사에게 선석 점유율이 높다 -> 오래기다려야된다. (대기시간이 증가한다)

은행창구를 놓고보면, 창구 직원이 쉬는시간 없이 일할수록(점유율이 높을수록) 은행입장에서는 효율적이지만, 손님이 조금만 몰려도 대기줄이 길게 늘어난다.

점유율이 60%에서 70%로 오를때보다 80%에서 90%로 오를때 대기가 엄청 크게 증가한다.
이러한 성질이 대기행렬의 일반적인 성질이다. 어느 한 이해관계자의 KPI만 올리는 서비스가 아니라 종합적으로 살펴볼수있는 서비스를 기획하는것이 바람직하다.

### 영향력 관심도 그리드

이해관계자가 여럿이면 모두에게 같은 노력을 들일 수 없다. 우선순위를 결정해야 한다. 이때 영향도 관심도 그리드(행렬)를 사용해볼 수 있다

가로축에 서비스에대한 관심도 세로축에 서비스 결정에 미치는 영향력을 두고 이해관계자를 네칸 가운데 하나에 넣고 칸마다 다른 관리 전략 쓴다.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
# 입항 시간 조정 서비스기준 채점결과

stakeholder = pd.DataFrame({
    '이해관계자': ['선사', '터미널운영사', '부산항만공사', '관제', '화주/포워더', '육상운송사', '예/도선업체'],
    '영향력': [5,4,5,4,2,1,2,],
    '관심도': [5,5,4,2,4,4,2,]
})

def manage_strategy(row):
    if row['영향력'] >= 3 and row['관심도'] >= 3:
        return '적극관리'
    if row['영향력'] >= 3:
        return '만족 유지'
    if row['관심도'] >= 3:
        return '정보 공유'
    return '관찰'

stakeholder['관리전략'] = stakeholder.apply(manage_strategy, axis=1)

stakeholder
```

```python
stakeholder.groupby('관리전략')['이해관계자'].apply(list).to_dict()
```

# 페르소나(Persona)

이해관계자 표는 선사, 화주처럼 집단을 다룬다. 그런데 서비스 화면 하나, 알림 문구 하나를 정할때 선사는 어떻게 이야기할까? 라고 물어보면 대답하기가 어렵다. 선사 안에서도 본선 스케줄 짜는 담당자, 입항신고 담당자, 비용 결제 담당자가 다 다르고, 원하는게 다르기 때문이다.

페르소나는 집단 안에서 서비스를 실제로 쓸 대표 사용자를 이름, 역할, 목표, 불편(Pain)을 갖는 가상의 한 사람으로 그린것.

## 구성요소

- 이름/역할: 가상의 이름, 소속, 직무
- 목표(Goal): 이사람이 일에서 성취하려는 것
- 불편(Pain Point): 목표를 가로막는 것
- 하루 업무: 언제 무엇을 하는지
- 쓰는 정보/도구: 판단에 쓰이는 자료
- 판단 KPI: 이 사람이 성과로 보는 숫자
- 근거: 각각의 내용에 대한 근거

## 구분

처음 페르소나를 만들때는 대부분 기획팀의 경험과 짐작으로 채워진다. 이렇게 경험과 짐작으로 만들어진 페르소나를 가정 페르소나(Proto-Persona)라고 일컫는다.

가정 페르소나는 기획을 빠르게 시작하는데에는 도움이 될 수 있지만 이것을 근거로 사용하면 문제가 될 수 있다. 인터뷰, 데이터와같은 것들을 하나씩 확인해서 다듬어 나가면 근거 페르소나(Evidence-based Persona)라고 일컫는다. 서비스 기획을 할떄에는 근거 페르소나를 사용하는것이 좋다.

- 누헨진(선사): 입항하는 업체는 몇곳이고, 몇몇 업체에 몰려있는가 -> 업체명/선박종류명
- 전준누(터미널): 화물이 어느 권역에서 처리되는가 -> 청코드
- 황성빙(포워더): 국내 화주가 있는 화물은 전체의 얼마부분인가 -> 수출입구분명/화물종류
- 레이네스(항만공사): 부산항에 들어오는 배는 무엇을 하러 오는가 -> 입항목적

## 주의할 부분

페르소나는 한번 만들고 쭉 사용하는것이 아니라 계속 고쳐나가야 한다. 문제정의서와 마찬가지로 정보가 갱신될대마다 고쳐지는 살아있는 문서라고 볼 수 있기 때문이다. '가정'한 부분은 반드시 실제 인터뷰나 데이터로 확인된 결과가 있으면 고쳐야한다.

- 고정관념 페르소나: 기획팀의 선입견을 사람 모양으로 포장 -> 근거가 모두 가정인 페르소나
- 평균 사용자: 여러 집단의 평균을 한 사람으로 만들면 실제로는 아무도 대표하지 못한다
- 대리지표 과신: 데이터의 이름과 실제 뜻이 다르게 들어가는경우
- 페르소나 과다: 페르소나가 너무 많으면 누구를 위해 설계하는지 흐려지는 문제
- 빠진 이해관계자: 데이터에 없는 집단은 페르소나에서도 빠지기 쉽다.
