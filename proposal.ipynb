# 개선안 구조

학교앞 횡단보도에서 아침에 학생들이 붐빈다. (관찰)

- 보행신호가 20초로 짧기 때문에 한번에 다 못건넌다. (원인 짐작)
- 아침 7시30분 ~ 8시30분 에는 신호를 한시적으로 40초로 늘리는것이 좋을것 같다. (해결방안)
- 한 번에 못건너고 남는 학생이 절반으로 줄어들 것이다. (기대효과)

관찰에서 건의까지 올라가는동안 확인된 사실보다 판단이 많아질 수 있다.

개선기회를 개선안으로 바꾸는 방법에 대해서 살펴보겠다.

## 사실에서 제안까지의 흐름

개선안 4가지 정도로 구성이 되어있다.

- 사실: 데이터로 확인한 사실
- 원인가설: 사실이 왜 생겼는지에 대한 가설
- 해결방안: 원인을 건드리는 해결 방안
- 기대효과: 해결이 실행되면 달라질 기대효과

위층으로 올라갈수록 아직 일어나지 않은 일에대한 판단이 많아지기때문에, 같은 확신정도로 이야기하기 어렵다.

- 사실: 제한이 풀린 어떤 기간에 가장 바쁜 한시간의 입항 척수가 13척이었다.(평소에는 10척이 들어온다)
- 원인가설: 통제가 풀리면 기다리던 선박의 입항요청이 한꺼번에 들어오는 경향이 있다.
- 해결방안: 통제 해제전에 입항 순서를 미리 배정한다.
- 기대효과: 가장 바쁜 한 시간대의 입항 척수가 평소 수준으로 낮아질것이다.

이런 구조를 한방향으로만 작성하게되면 마치 해결방안이 처음부터 하나로 정해져 있는것처럼 보일 수 있다.
원인 가설 윗층에 해당되는 부분은 여러개의 내용들이 올라올 수 있다.

해결방안에 해당되는 부분은 한가지 내용이 아니라 여러개의 해결방안이 올라올수있다.

- 사실: 제한이 풀린 어떤 기간에 가장 바쁜 한시간의 입항 척수가 13척이었다.(평소에는 10척이 들어온다)
- 원인가설: 통제가 풀리면 기다리던 선박의 입항요청이 한꺼번에 들어오는 경향이 있다.
- 해결방안:
  - 1. 통제 해제전에 입항 순서를 미리 배정한다.
    2. 해제 예상 시각을 사전에 공지해서 스케줄을 구성할 수 있도록 조치한다.
    3. 예선 예상 소요시간을 사전에 통보하여 스케줄을 구성할 수 있도록 조치한다.
- 기대효과: 가장 바쁜 한 시간대의 입항 척수가 평소 수준으로 낮아질것이다.
