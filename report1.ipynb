# 보고서 자동화

군집, 패턴, 이상징후, 예측과 같은 분석도구를 사용할 수 있게 되었다. 우리의 분석도구가 대시보드안에만, 쥬피터 노트북 파일 안에만 들어있으면 의사결정에 사용할 수 없다.

## 분석 결과와 보고서

분석 자료(결과)는 분석가가 무엇이 사실인가를 찾아가는 작업 기록인 것이고, 분석 보고서는 의사결정자가 그래서 뭉서을 해야하는가를 판단하도록 돕는 문서이다.

분석 보고서에서는 결론 -> 근거 -> 세부 형식(두괄식)으로 구성되어있어야 한다.

## 분석 보고서의 구성요소

- 제목과 기준기간 : 부산항 컨테이너 물동량 월간보고(2026년 10월)
- 핵심지표 카드(KPI): 당월 물동량, 전년대비 동월 증감률, 환적비중
- 요약 문장: 전년 동월보다 4.5%p 증가했습니다.
- 차트: 최근 25개월 분량의 월별 물동량 추이 도표
- 주석: 단위 TEU, 기록방법 수기

### 수기 보고서와의 차이

- 옮겨적는데 발생하는 오류가 적어짐
- 도표 재사용성 증가
- 기간 혼동을 방지
- 계산 오류도 어느정도 예방
- 단위 혼동 예방가능

## 보고서 차트 선택 기준

탐색 단계에서의 도표(차트)는 무엇이 있는지를 찾으려고 여러개를 그리는것이 일반적이다. 보고서의 차트는 이미 정한 메세지 하나만을 보고 대상에게 전달하는 것이다. 보고서에 작성되는 도표는 먼저 이 도표로 어떤 질문에 답하려는것인가를 확정하고, 그 질문에 맞는 유형을 선택해야 한다.

- 시간에 따라 어떻게 변했는가(추이): 선그래프
- 무엇이 더 큰것인가(비교): 막대그래프
- 전체가어떻게 나뉘는가(구성): 누적 막대그래프
- 얼마나 늘고 줄었나(변화): 위아래로 그려지는 막대그래프

## 숫자 결과를 문장으로 표현

보고서안에 작성되는 숫자는 계산결과를 그대로 포함시키는것보다는, 보고대상자가 읽기 쉬운 단위, 모양으로 변경해서 작성하는것이 좋다. 24,402,022 TEU보다는 2440만 TEU, 5.3911%보다는 5.4%가 적합하다.

카드에서는 4.4%, 본문에서는 4.48%, 표에서는 4.5% 일정한 기준이필요하다

### 반올림과 표시단위

원래 값에서 '한번만 반올림'해야한다. 이미 반올림한 값을 다시한번 반올림하면 값에오류가 발생할수 있기 때문이다.

구성비 33.35, 33.35, 33.30이 있다고 하면, 이 값들을 다시 반올림하게되면, 33.4, 33.4, 33.3, 총합은 100.1.

파이썬에서 반올림할때에는 `round()`를 사용하는데, 정혹히 가운데값인 5를 가장 가까운 짝수 쪽으로 보내는 반올림 방법을 사용하는데, 은행가 반올림 법칙
을 사용한다. 파이썬을 이용해서 반올림같은 기능을 사용할때에는 decimal과 같은 정밀한 자료형을 사용하는것이 권장된다.

은행가 반올림을 사요하는 이유는 가운데 값을 항상 올리게되면 많은 숫자를 반올림해서 더할떄 합계가 조금씩 위로 치우쳐지기 때문이다.


```python
round(2.5)
```
